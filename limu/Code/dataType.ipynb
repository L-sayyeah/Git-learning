## 张量连结

```python
x = torch.arange(12, dtype=torch.float32).reshape((3, 4))
y = torch.tensor([[2.0, 1, 4, 3], [2.0, 1, 4, 3], [2.0, 1, 4, 3]])
# 竖向连结(行)        # 横向连结(列)
torch.cat((x, y), dim=0), torch.cat((x,y), dim=1)
# 三维 dim = 2等等
```

## 运行一些操作可能导致为新结果分配内存

```python
before = id(y)
y = y + x
id(y) == before
# y 不会等于之前的 y,因为在y重新赋值的时候,之前的y已经被析构了
```

## 原地操作

```python
z = torch.zeros_like(y)
print("id:z", id(z))
z[:] = x + y
print("id:z", id(z))
```

## 如果后续计算中没有重复使用x, 也可以使用x[:] = x+y或者 x+=y来减少操作的内存开销

```python
before = id(x)
x += y
id(x) == before
```
