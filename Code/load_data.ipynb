```python
import os
#将某个文件夹下的所有文件名做一个列表
dir_path = "./datasets/hymenoptera_data/train/ants"
img_path_list = os.listdir(dir_path)
img_path_list[0]
```

```python
#将两个路径相加
root_dir = "./datasets/hymenoptera_data/train"
label_dir = "ants"

path = os.path.join(root_dir, label_dir)
img_name = img_path_list[0]
img_path = os.path.join(root_dir, label_dir, img_name)
img_path
```

```python
#Dataset类实战
class MyData(Dataset):
    
    def __init__(self,root_dir,label_dir):
        self.root_dir = root_dir
        self.label_dir = label_dir
        self.path = os.path.join(self.root_dir, self.label_dir)
        self.img_path = os.listdir(self.path)
        
    def __getitem__(self, idx):
        img_name = self.img_path[idx]
        img_item_path = os.path.join(self.root_dir, self.label_dir, img_name)
        img = Image.open(img_item_path)
        label = self.label_dir
        return img,label

    def __len__(self):
        return len(self.img_path)
```

```python
#蚂蚁数据集
ants_dataset = MyData("./datasets/hymenoptera_data/train", "ants")
img,label = ants_dataset[0]
print(img,"\n",label)
img.show()
print(len(ants_dataset))
```

```python
#蜜蜂数据集
bees_dataset = MyData("./datasets/hymenoptera_data/train", "bees")
img,label = bees_dataset[0]
print(img,"\n",label)
img.show()
print(len(bees_dataset))
```

```python
#数据集拼接，直接相加
train_dataset = ants_dataset + bees_dataset
print(len(train_dataset))
```
