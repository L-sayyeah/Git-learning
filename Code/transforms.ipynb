```python
from torchvision import transforms 
from PIL import Image
import cv2
from torch.utils.tensorboard import SummaryWriter
```

```python
help(transforms)
```

```python
#tensor数据类型
#通过transforms.ToTensor
#1.transforms如何使用
#2.tensor与其他类型的区别，为什么要用tensor
help(transforms.ToTensor)
```

```python
#1.transforms如何使用
img_path = "./datasets/hymenoptera_data/train/ants/0013035.jpg"
img = Image.open(img_path)
print(img)
tensor_trans = transforms.ToTensor()
tensor_img = tensor_trans(img)
print(tensor_img)
```

```python
#为什么需要tensor类型:tensor包括了神经网络训练过程中的一些参数
print(tensor_img.type())
```

```python
#opencv读入img，是numpy
cv_img = cv2.imread(img_path)

writer = SummaryWriter("logs")

writer.add_image("test1",tensor_img)
writer.close()
```
